```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
```

```python
df=pd.read_csv("price_of_healthy_diet_clean.csv")
df.head()
```

```python
df.isnull().sum()
df.drop_duplicates(inplace=True)
```

```python
region_cost = df.groupby("region")["cost_healthy_diet_ppp_usd"].mean().reset_index()
region_cost
sns.barplot(data=region_cost, x="region", y="cost_healthy_diet_ppp_usd")
plt.title("Average Cost of a Healthy Diet by Region")
plt.xticks(rotation=45)
plt.show()
```

```python
top_countries = (
    df.groupby("country")["cost_healthy_diet_ppp_usd"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
top_countries
sns.barplot(data=top_countries, x="cost_healthy_diet_ppp_usd", y="country")
plt.title("Top 10 Countries with Highest Healthy Diet Cost")
plt.show()
```

```python
income_cost = df.groupby("cost_category")["cost_healthy_diet_ppp_usd"].mean().reset_index()
income_cost
sns.barplot(data=income_cost, x="cost_category", y="cost_healthy_diet_ppp_usd")
plt.title("Healthy Diet Cost by Cost Category")
plt.show()
```

```python
year_cost = df.groupby("year")["cost_healthy_diet_ppp_usd"].mean().reset_index()
sns.lineplot(data=year_cost, x="year", y="cost_healthy_diet_ppp_usd")
plt.title("Trend of Healthy Diet Cost Over Time")
plt.show()
```
